```python
# ============================================================
#  ETAPE 2 : Chargement du moteur + interface
# ============================================================
from tabulate import tabulate
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import seaborn as sns
import random, time
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Palette de couleurs
C = dict(
    bg='#0f1117', card='#1a1d2e', border='#2a2d3e',
    accent='#6c63ff', accent2='#ff6584',
    green='#43d98c', yellow='#ffd166',
    text='#e8eaf6', muted='#8b8fa8',
)

plt.rcParams.update({
    'figure.facecolor': C['card'], 'axes.facecolor': C['card'],
    'axes.edgecolor': C['border'], 'axes.labelcolor': C['text'],
    'xtick.color': C['muted'], 'ytick.color': C['muted'],
    'text.color': C['text'], 'grid.color': C['border'],
    'grid.linewidth': 0.6, 'font.family': 'DejaVu Sans',
    'axes.spines.top': False, 'axes.spines.right': False,
})

CSS = """
<style>
@import url('https://fonts.googleapis.com/css2?family=Space+Mono:wght@400;700&family=DM+Sans:wght@300;400;600;700&display=swap');
.sda-app{font-family:'DM Sans',sans-serif;color:#e8eaf6;}
.sda-hero{
  background:linear-gradient(135deg,#0f1117 0%,#1a1d2e 40%,#12102b 100%);
  border:1px solid #2a2d3e;border-radius:16px;padding:40px 48px;
  margin-bottom:24px;position:relative;overflow:hidden;
}
.sda-hero::before{
  content:'';position:absolute;top:-60px;right:-60px;
  width:220px;height:220px;border-radius:50%;
  background:radial-gradient(circle,rgba(108,99,255,0.25) 0%,transparent 70%);
  pointer-events:none;
}
.sda-hero-title{
  font-family:'Space Mono',monospace;font-size:26px;font-weight:700;
  background:linear-gradient(90deg,#6c63ff,#ff6584);
  -webkit-background-clip:text;-webkit-text-fill-color:transparent;
  margin:0 0 6px 0;
}
.sda-hero-sub{font-size:13px;color:#8b8fa8;margin:0 0 18px 0;}
.sda-badge{
  display:inline-block;
  background:rgba(108,99,255,0.15);border:1px solid rgba(108,99,255,0.4);
  color:#a29bfe;border-radius:20px;padding:3px 12px;
  font-size:11px;margin-right:6px;margin-top:4px;
  font-family:'Space Mono',monospace;
}
.sda-section{
  background:#1a1d2e;border:1px solid #2a2d3e;
  border-left:4px solid #6c63ff;border-radius:12px;
  padding:20px 28px;margin-bottom:16px;
}
.sda-section-title{
  font-family:'Space Mono',monospace;font-size:12px;font-weight:700;
  color:#6c63ff;letter-spacing:2px;text-transform:uppercase;margin:0 0 14px 0;
}
.sda-stats{
  display:grid;grid-template-columns:repeat(auto-fit,minmax(140px,1fr));
  gap:12px;margin-bottom:16px;
}
.sda-stat{
  background:#12102b;border:1px solid #2a2d3e;
  border-radius:12px;padding:16px 20px;text-align:center;
}
.sda-stat-val{font-family:'Space Mono',monospace;font-size:28px;font-weight:700;color:#6c63ff;line-height:1;margin-bottom:4px;}
.sda-stat-val.g{color:#43d98c;}.sda-stat-val.y{color:#ffd166;}.sda-stat-val.p{color:#ff6584;}
.sda-stat-lbl{font-size:11px;color:#8b8fa8;text-transform:uppercase;letter-spacing:1px;}
.sda-table{width:100%;border-collapse:collapse;font-size:13px;margin-top:8px;}
.sda-table th{
  background:rgba(108,99,255,0.15);color:#a29bfe;font-weight:600;
  padding:10px 16px;text-align:left;font-family:'Space Mono',monospace;
  font-size:11px;letter-spacing:1px;text-transform:uppercase;
  border-bottom:1px solid #2a2d3e;
}
.sda-table td{padding:10px 16px;color:#e8eaf6;border-bottom:1px solid rgba(42,45,62,0.5);}
.sda-table tr:last-child td{border-bottom:none;}
.sda-table tr:hover td{background:rgba(108,99,255,0.07);}
.chip-y{display:inline-block;background:rgba(67,217,140,0.15);border:1px solid rgba(67,217,140,0.4);color:#43d98c;border-radius:20px;padding:2px 10px;font-size:11px;}
.chip-n{display:inline-block;background:rgba(255,101,132,0.12);border:1px solid rgba(255,101,132,0.3);color:#ff6584;border-radius:20px;padding:2px 10px;font-size:11px;}
.sda-bar-wrap{background:#12102b;border-radius:20px;height:10px;margin-top:8px;overflow:hidden;border:1px solid #2a2d3e;}
.sda-bar-fill{height:100%;border-radius:20px;background:linear-gradient(90deg,#6c63ff,#43d98c);}
.sda-bar-lbl{display:flex;justify-content:space-between;font-size:12px;color:#8b8fa8;margin-top:4px;}
.sda-compare{display:grid;grid-template-columns:1fr 1fr;gap:12px;margin-top:8px;}
.sda-ccard{background:#12102b;border:1px solid #2a2d3e;border-radius:10px;padding:16px 20px;}
.sda-ccard.a{border-color:#43d98c;}
.sda-ctitle{font-family:'Space Mono',monospace;font-size:12px;color:#6c63ff;margin-bottom:8px;}
.sda-ccard.a .sda-ctitle{color:#43d98c;}
.sda-crow{display:flex;justify-content:space-between;font-size:12px;color:#8b8fa8;margin:4px 0;}
.sda-crow span{color:#e8eaf6;}
.sda-trace{background:#12102b;border:1px solid #2a2d3e;border-radius:10px;padding:16px 20px;font-family:'Space Mono',monospace;font-size:12px;line-height:2;margin-top:8px;}
.tr-y{color:#43d98c;}.tr-n{color:#8b8fa8;}.tr-a{color:#6c63ff;}
.sda-trow{display:flex;align-items:center;padding:8px 0;border-bottom:1px solid #2a2d3e;font-size:13px;}
.sda-trow:last-child{border-bottom:none;}
.sda-tname{flex:1;color:#e8eaf6;}
.sda-tparam{color:#8b8fa8;font-size:11px;font-family:'Space Mono',monospace;margin-right:16px;}
.t-ok{color:#43d98c;font-weight:700;}.t-fail{color:#ff6584;font-weight:700;}
</style>
"""

def inject_css():
    display(HTML(CSS))

# ── Algorithmes ──────────────────────────────────────────────
def sac_a_dos_01(poids, valeurs, capacite):
    n = len(poids)
    if len(valeurs) != n: raise ValueError("Listes de tailles differentes")
    if any(p < 0 for p in poids) or any(v < 0 for v in valeurs): raise ValueError("Valeurs negatives")
    if capacite < 0: raise ValueError("Capacite negative")
    DP = [[0]*(capacite+1) for _ in range(n+1)]
    for i in range(1, n+1):
        for w in range(capacite+1):
            if poids[i-1] <= w:
                DP[i][w] = max(DP[i-1][w], DP[i-1][w-poids[i-1]]+valeurs[i-1])
            else:
                DP[i][w] = DP[i-1][w]
    w = capacite; sel = []; pt = 0
    for i in range(n, 0, -1):
        if DP[i][w] != DP[i-1][w]:
            sel.append(i-1); pt += poids[i-1]; w -= poids[i-1]
    sel.reverse()
    return DP[n][capacite], sel, pt, DP

def sac_a_dos_1D(poids, valeurs, capacite):
    dp = [0]*(capacite+1)
    for i in range(len(poids)):
        for w in range(capacite, poids[i]-1, -1):
            dp[w] = max(dp[w], dp[w-poids[i]]+valeurs[i])
    return dp[capacite]

# ── Rendu solution HTML ──────────────────────────────────────
def render_solution(poids, valeurs, capacite, noms):
    val, sel_idx, pt, DP = sac_a_dos_01(poids, valeurs, capacite)
    n = len(poids); sel_set = set(sel_idx)
    pct = round(pt/capacite*100,1) if capacite > 0 else 0
    val_1d = sac_a_dos_1D(poids, valeurs, capacite)
    mem_2d = (n+1)*(capacite+1); mem_1d = capacite+1

    def stat(v,l,c=''):
        return f'<div class="sda-stat"><div class="sda-stat-val {c}">{v}</div><div class="sda-stat-lbl">{l}</div></div>'

    stats_html = f"""
    <div class="sda-stats">
      {stat(val, 'Valeur optimale','g')}
      {stat(pt, 'Poids total','y')}
      {stat(len(sel_idx), 'Objets selectionnes','')}
      {stat(f'{pct}%', 'Capacite utilisee','p')}
    </div>
    <div class="sda-bar-wrap"><div class="sda-bar-fill" style="width:{pct}%"></div></div>
    <div class="sda-bar-lbl"><span>Capacite utilisee</span><span>{pt}/{capacite} ({pct}%)</span></div>
    """

    rows = ''; cp = cv = 0
    for i in range(n):
        is_sel = i in sel_set
        chip = '<span class="chip-y">Pris</span>' if is_sel else '<span class="chip-n">Laisse</span>'
        ratio = round(valeurs[i]/poids[i],2) if poids[i]>0 else 'inf'
        if is_sel: cp += poids[i]; cv += valeurs[i]
        rows += f"<tr><td><b>{noms[i]}</b></td><td>{poids[i]}</td><td>{valeurs[i]}</td><td>{ratio}</td><td>{chip}</td><td style='color:#43d98c'>{cv if is_sel else '-'}</td></tr>"

    table_html = f"""
    <table class="sda-table">
      <thead><tr><th>Nom</th><th>Poids</th><th>Valeur</th><th>Ratio V/P</th><th>Selection</th><th>Cumul valeur</th></tr></thead>
      <tbody>{rows}</tbody>
    </table>"""

    w_tr = capacite; trace = ''
    for i in range(n, 0, -1):
        pris = DP[i][w_tr] != DP[i-1][w_tr]
        if pris:
            trace += f'<div class="tr-y">i={i} ({noms[i-1]}) DP[{i}][{w_tr}]={DP[i][w_tr]} <span class="tr-a">-></span> PRIS  (w: {w_tr}->{w_tr-poids[i-1]})</div>'
            w_tr -= poids[i-1]
        else:
            trace += f'<div class="tr-n">i={i} ({noms[i-1]}) DP[{i}][{w_tr}]={DP[i][w_tr]} -> ignore</div>'
    trace += f'<br><span class="tr-y">Valeur optimale : {val}</span>'

    compare_html = f"""
    <div class="sda-compare">
      <div class="sda-ccard">
        <div class="sda-ctitle">DP 2D - Classique</div>
        <div class="sda-crow">Memoire <span>O(n x W)</span></div>
        <div class="sda-crow">Cases <span>{mem_2d:,}</span></div>
        <div class="sda-crow">Valeur <span>{val}</span></div>
        <div class="sda-crow">Reconstruction <span style='color:#43d98c'>Possible</span></div>
      </div>
      <div class="sda-ccard a">
        <div class="sda-ctitle">DP 1D - Optimise</div>
        <div class="sda-crow">Memoire <span>O(W)</span></div>
        <div class="sda-crow">Cases <span style='color:#43d98c'>{mem_1d:,} (x{mem_2d//max(mem_1d,1)} moins)</span></div>
        <div class="sda-crow">Valeur <span>{val_1d}</span></div>
        <div class="sda-crow">Reconstruction <span style='color:#ff6584'>Non disponible</span></div>
      </div>
    </div>"""

    hero = f"""
    <div class="sda-app">
    <div class="sda-hero">
      <div class="sda-hero-title">Sac a Dos 0/1 - Solution Optimale</div>
      <div class="sda-hero-sub">Programmation Dynamique - Master ISI</div>
      <span class="sda-badge">O(n x W)</span>
      <span class="sda-badge">n = {n}</span>
      <span class="sda-badge">W = {capacite}</span>
    </div>
    <div class="sda-section"><div class="sda-section-title">Resume</div>{stats_html}</div>
    <div class="sda-section"><div class="sda-section-title">Objets</div>{table_html}</div>
    <div class="sda-section"><div class="sda-section-title">Trace du Backtracking</div><div class="sda-trace">{trace}</div></div>
    <div class="sda-section"><div class="sda-section-title">DP 2D vs DP 1D</div>{compare_html}</div>
    </div>"""

    display(HTML(hero))
    return val, sel_idx, pt, DP

# ── Graphiques ───────────────────────────────────────────────
def render_charts(poids, valeurs, sel_idx, capacite, noms):
    n = len(poids); sel_set = set(sel_idx)
    ratios = [v/p if p>0 else 0 for v,p in zip(valeurs,poids)]
    idx_list = list(range(n))
    short = [nm[:8] for nm in noms]

    fig, axes = plt.subplots(2, 2, figsize=(14, 9))
    fig.patch.set_facecolor(C['card'])
    fig.suptitle('Analyse Visuelle - Sac a Dos 0/1', fontsize=14, fontweight='bold', color=C['text'], y=1.01)

    def sax(ax, title):
        ax.set_title(title, fontsize=11, color=C['text'], pad=10)
        ax.grid(True, axis='y', alpha=0.2)
        ax.set_xticks(idx_list); ax.set_xticklabels(short, fontsize=9)
        for sp in ax.spines.values(): sp.set_edgecolor(C['border'])

    clrs = [C['green'] if i in sel_set else C['muted'] for i in idx_list]
    alpha = [1.0 if i in sel_set else 0.4 for i in idx_list]
    bars = axes[0,0].bar(idx_list, valeurs, color=clrs, edgecolor='none', width=0.6)
    for b,a in zip(bars,alpha): b.set_alpha(a)
    for b,v in zip(bars,valeurs):
        axes[0,0].text(b.get_x()+b.get_width()/2, b.get_height()+0.3, str(v), ha='center', va='bottom', fontsize=9, color=C['text'])
    sax(axes[0,0], 'Valeurs (vert = selectionne)')
    axes[0,0].set_ylabel('Valeur', color=C['muted'])

    clrs2 = [C['yellow'] if i in sel_set else C['muted'] for i in idx_list]
    bars2 = axes[0,1].bar(idx_list, poids, color=clrs2, edgecolor='none', width=0.6)
    for b,a in zip(bars2,alpha): b.set_alpha(a)
    axes[0,1].axhline(y=capacite, color=C['accent2'], linestyle='--', linewidth=1.8, label=f'Capacite max ({capacite})')
    axes[0,1].legend(fontsize=9, framealpha=0.2, labelcolor=C['text'])
    for b,p in zip(bars2,poids):
        axes[0,1].text(b.get_x()+b.get_width()/2, b.get_height()+0.1, str(p), ha='center', va='bottom', fontsize=9, color=C['text'])
    sax(axes[0,1], 'Poids'); axes[0,1].set_ylabel('Poids', color=C['muted'])

    clrs3 = [C['accent'] if i in sel_set else C['muted'] for i in idx_list]
    bars3 = axes[1,0].bar(idx_list, ratios, color=clrs3, edgecolor='none', width=0.6)
    for b,a in zip(bars3,alpha): b.set_alpha(a)
    for b,r in zip(bars3,ratios):
        axes[1,0].text(b.get_x()+b.get_width()/2, b.get_height()+0.02, f'{r:.2f}', ha='center', va='bottom', fontsize=9, color=C['text'])
    sax(axes[1,0], 'Ratio Valeur/Poids'); axes[1,0].set_ylabel('Ratio', color=C['muted'])

    sv = sum(valeurs[i] for i in idx_list if i in sel_set)
    nv = sum(valeurs[i] for i in idx_list if i not in sel_set)
    if sv+nv > 0:
        wedges,texts,autotexts = axes[1,1].pie(
            [sv, nv], labels=['Selectionnes','Non selectionnes'],
            colors=[C['accent'], C['border']],
            autopct='%1.0f%%', startangle=140,
            wedgeprops=dict(width=0.55, edgecolor=C['card'], linewidth=3),
            textprops=dict(color=C['text'], fontsize=10)
        )
        for at in autotexts: at.set_fontsize(11); at.set_color(C['text'])
    axes[1,1].set_title('Repartition des valeurs', fontsize=11, color=C['text'], pad=10)

    plt.tight_layout(); plt.show()

# ── Visualisation pas-a-pas ──────────────────────────────────
def render_stepbystep(poids, valeurs, capacite, noms):
    n = len(poids)
    DP = [[0]*(capacite+1) for _ in range(n+1)]
    for i in range(1, n+1):
        for w in range(capacite+1):
            if poids[i-1] <= w:
                DP[i][w] = max(DP[i-1][w], DP[i-1][w-poids[i-1]]+valeurs[i-1])
            else:
                DP[i][w] = DP[i-1][w]

    out = widgets.Output()

    # ── Parametres adaptatifs selon la taille de la table ──────
    cell_w   = max(0.30, min(0.65, 10.0 / (capacite + 2)))
    cell_h   = max(0.40, min(0.85,  6.0 / (n + 2)))
    fig_w    = min(26, max(14, (capacite + 2) * cell_w + 6))
    fig_h    = min(20, max(5,  (n + 2)        * cell_h + 2))
    ann_sz   = max(5, min(9, int(110 / (capacite + 2))))
    do_annot = (capacite <= 40) and (n <= 20)
    step_x   = max(1, (capacite + 1) // 20)
    xlabels  = [str(w) if w % step_x == 0 else '' for w in range(capacite + 1)]

    def draw(etape):
        with out:
            clear_output(wait=True)
            fig, axes = plt.subplots(1, 2, figsize=(fig_w, fig_h))
            fig.patch.set_facecolor(C['card'])

            data = np.array([[DP[i][w] if i<=etape else 0
                              for w in range(capacite+1)]
                             for i in range(n+1)], dtype=float)
            mask = np.array([[i>etape for _ in range(capacite+1)]
                             for i in range(n+1)])
            cmap = sns.light_palette('#6c63ff', as_cmap=True)

            sns.heatmap(data, annot=do_annot, fmt='.0f', cmap=cmap,
                        mask=mask, ax=axes[0],
                        linewidths=0.3, linecolor='#12102b',
                        xticklabels=xlabels,
                        yticklabels=[f'i={i}' for i in range(n+1)],
                        annot_kws={'size': ann_sz}, cbar=False)

            # Zones non encore remplies
            for i in range(etape+1, n+1):
                for w in range(capacite+1):
                    axes[0].add_patch(plt.Rectangle(
                        (w, i), 1, 1, facecolor='#12102b', alpha=0.75))
            # Surligner la ligne courante
            if etape > 0:
                for w in range(capacite+1):
                    axes[0].add_patch(plt.Rectangle(
                        (w, etape), 1, 1, fill=False,
                        edgecolor=C['accent2'], linewidth=2.0))

            axes[0].set_title(f'Table DP - Etape {etape}/{n}',
                              fontsize=12, color=C['text'], pad=10)
            axes[0].tick_params(colors=C['muted'], labelsize=8)
            for sp in axes[0].spines.values():
                sp.set_edgecolor(C['border'])

            # ── Panneau droit : infos ───────────────────────────
            axes[1].set_facecolor(C['card']); axes[1].axis('off')
            if etape == 0:
                info = [
                    ('Etape',  '0 / ' + str(n),          C['accent']),
                    ('Action', 'Initialisation',           C['text']),
                    ('Regle',  'DP[0][w]=0 pour tout w',  C['muted']),
                ]
            else:
                obj = etape - 1
                info = [
                    ('Etape',      f'{etape} / {n}',                 C['accent']),
                    ('Objet',      noms[obj],                         C['green']),
                    ('Poids',      str(poids[obj]),                   C['yellow']),
                    ('Valeur',     str(valeurs[obj]),                  C['text']),
                    ('Ratio',      f'{valeurs[obj]/poids[obj]:.2f}',  C['muted']),
                    ('Regle',      'si poids[i]<=w :',                C['muted']),
                    ('',           'max(DP[i-1][w],',                 C['muted']),
                    ('',           'DP[i-1][w-p]+v)',                 C['muted']),
                    ('Max actuel', str(DP[etape][capacite]),          C['green']),
                ]

            # ── CORRECTION : on ne passe jamais col='' a matplotlib ──
            y = 0.95
            for label, val, col in info:
                safe_col = col if col else C['text']
                if label:
                    axes[1].text(0.05, y, f'{label}:',
                                 transform=axes[1].transAxes,
                                 fontsize=11, color=C['muted'],
                                 va='top', fontweight='bold')
                if val:
                    axes[1].text(0.42, y, val,
                                 transform=axes[1].transAxes,
                                 fontsize=11, color=safe_col, va='top')
                y -= 0.08

            if etape == n:
                axes[1].text(0.05, y - 0.06,
                             f'TERMINE\nValeur optimale : {DP[n][capacite]}',
                             transform=axes[1].transAxes, fontsize=13,
                             color=C['green'], va='top', fontweight='bold',
                             bbox=dict(boxstyle='round,pad=0.5',
                                       facecolor='#0f3328',
                                       edgecolor=C['green'], alpha=0.9))

            # Message si annotations desactivees
            if not do_annot:
                axes[0].text(0.5, -0.06,
                             'Valeurs non affichees (table trop grande - capacite > 40)',
                             transform=axes[0].transAxes, fontsize=9,
                             color=C['muted'], ha='center')

            plt.tight_layout(); plt.show()

    slider = widgets.IntSlider(
        value=0, min=0, max=n, step=1, description='Etape :',
        style={'description_width': 'initial', 'handle_color': '#6c63ff'},
        layout=widgets.Layout(width='65%'))
    btn_n = widgets.Button(description='Suivant', button_style='primary',
                           layout=widgets.Layout(width='100px'))
    btn_r = widgets.Button(description='Reset',   button_style='warning',
                           layout=widgets.Layout(width='90px'))
    btn_n.on_click(lambda b: setattr(slider, 'value', min(slider.value+1, n)))
    btn_r.on_click(lambda b: setattr(slider, 'value', 0))
    widgets.interactive_output(draw, {'etape': slider})

    display(HTML('<div style="color:#8b8fa8;font-size:13px;margin-bottom:8px;">'
                 'Deplacez le slider ou cliquez Suivant pour avancer objet par objet</div>'))
    display(widgets.HBox([slider, btn_n, btn_r],
                         layout=widgets.Layout(align_items='center', margin='0 0 10px 0')))
    display(out)
    draw(0)


# ── Heatmap Backtracking ─────────────────────────────────────
def render_backtracking(poids, valeurs, capacite, noms):
    n = len(poids)
    DP = [[0]*(capacite+1) for _ in range(n+1)]
    for i in range(1, n+1):
        for w in range(capacite+1):
            if poids[i-1] <= w:
                DP[i][w] = max(DP[i-1][w], DP[i-1][w-poids[i-1]]+valeurs[i-1])
            else:
                DP[i][w] = DP[i-1][w]

    chemin = []; prises = []; w = capacite
    for i in range(n, 0, -1):
        chemin.append((i, w))
        if DP[i][w] != DP[i-1][w]:
            prises.append((i, w)); w -= poids[i-1]
    chemin.append((0, w))

    DP_np   = np.array(DP, dtype=float)
    overlay = np.zeros((n+1, capacite+1))
    for (i, ww) in chemin:  overlay[i][ww] = 1
    for (i, ww) in prises:  overlay[i][ww] = 2

    ann_sz   = max(5, min(9, int(110 / (capacite + 2))))
    do_annot = (capacite <= 40) and (n <= 20)
    step_x   = max(1, (capacite + 1) // 20)
    xlabels  = [str(w) if w % step_x == 0 else '' for w in range(capacite + 1)]

    fig, axes = plt.subplots(1, 2, figsize=(16, max(5, (n+1)*0.7+2)))
    fig.patch.set_facecolor(C['card'])
    fig.suptitle('Table DP avec chemin de backtracking',
                 fontsize=13, fontweight='bold', color=C['text'])

    ylabels = ['init'] + [f'{noms[i]}\n(p={poids[i]}, v={valeurs[i]})' for i in range(n)]
    cmap = sns.light_palette('#6c63ff', as_cmap=True)
    sns.heatmap(DP_np, annot=do_annot, fmt='.0f', cmap=cmap, ax=axes[0],
                linewidths=0.3, linecolor='#12102b',
                xticklabels=xlabels, yticklabels=ylabels,
                annot_kws={'size': ann_sz},
                cbar_kws={'label': 'Valeur DP', 'shrink': 0.8})

    for i in range(n+1):
        for ww in range(capacite+1):
            if overlay[i][ww] == 2:
                axes[0].add_patch(plt.Rectangle(
                    (ww,i),1,1,fill=True,facecolor=C['green'],alpha=0.55,zorder=3))
                axes[0].add_patch(plt.Rectangle(
                    (ww,i),1,1,fill=False,edgecolor=C['green'],linewidth=3,zorder=4))
            elif overlay[i][ww] == 1:
                axes[0].add_patch(plt.Rectangle(
                    (ww,i),1,1,fill=False,edgecolor=C['accent2'],linewidth=2.5,zorder=4))

    axes[0].set_title('Chemin : rouge=visite | vert=selectionne',
                      fontsize=10, color=C['muted'])
    axes[0].set_xlabel('Capacite w', color=C['muted'])
    axes[0].set_ylabel('Objets', color=C['muted'])
    axes[0].tick_params(colors=C['muted'], labelsize=8)
    for sp in axes[0].spines.values(): sp.set_edgecolor(C['border'])

    axes[1].set_facecolor(C['card']); axes[1].axis('off')
    axes[1].set_title('Trace du backtracking',
                      fontsize=12, color=C['text'], fontweight='bold')

    w_trace = capacite
    lines_bt = [f'Depart : w = {capacite}', '']
    for i in range(n, 0, -1):
        pris = DP[i][w_trace] != DP[i-1][w_trace]
        sym  = 'PRIS' if pris else 'ignore'
        lines_bt.append(f'i={i} ({noms[i-1]}) DP[{i}][{w_trace}]={DP[i][w_trace]}  {sym}')
        if pris:
            lines_bt.append(f'   w : {w_trace} -> {w_trace-poids[i-1]}')
            w_trace -= poids[i-1]
    lines_bt.append(f'\nValeur optimale : {DP[n][capacite]}')

    axes[1].text(0.05, 0.95, '\n'.join(lines_bt),
                 transform=axes[1].transAxes,
                 fontsize=10, va='top', fontfamily='monospace', color=C['text'],
                 bbox=dict(boxstyle='round', facecolor='#12102b',
                           edgecolor=C['accent'], alpha=0.9))

    p_sel = mpatches.Patch(color=C['green'], alpha=0.6, label='Objet selectionne')
    p_ch  = mpatches.Patch(color='white', ec=C['accent2'], linewidth=2, label='Cellule visitee')
    axes[0].legend(handles=[p_sel, p_ch], loc='upper left', fontsize=9,
                   framealpha=0.2, labelcolor=C['text'])
    plt.tight_layout(); plt.show()

# ── Benchmark ────────────────────────────────────────────────
def render_benchmark(W=50, n_max=20):
    print('Benchmark en cours...')
    def force_brute(p, v, cap):
        n = len(p); best = 0
        for m in range(1 << n):
            pt = sum(p[i] for i in range(n) if m & (1 << i))
            vt = sum(v[i] for i in range(n) if m & (1 << i))
            if pt <= cap: best = max(best, vt)
        return best

    ns_dp = []; t_dp = []
    for n in range(5, 201, 10):
        p = [random.randint(1,10) for _ in range(n)]
        v = [random.randint(1,20) for _ in range(n)]
        t0 = time.perf_counter(); sac_a_dos_01(p, v, W)
        t_dp.append((time.perf_counter()-t0)*1000); ns_dp.append(n)

    ns_cmp = list(range(5, n_max+1)); t_dp_cmp = []; t_fb_cmp = []
    for n in ns_cmp:
        p = [random.randint(1,5) for _ in range(n)]
        v = [random.randint(1,10) for _ in range(n)]
        t0 = time.perf_counter(); sac_a_dos_01(p, v, W)
        t_dp_cmp.append((time.perf_counter()-t0)*1000)
        t0 = time.perf_counter(); force_brute(p, v, W)
        t_fb_cmp.append((time.perf_counter()-t0)*1000)

    Ws = []; t_W = []
    p_fix = [random.randint(1,10) for _ in range(20)]
    v_fix = [random.randint(1,20) for _ in range(20)]
    for cap in range(10, 501, 20):
        t0 = time.perf_counter(); sac_a_dos_01(p_fix, v_fix, cap)
        t_W.append((time.perf_counter()-t0)*1000); Ws.append(cap)

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.patch.set_facecolor(C['card'])
    fig.suptitle('Benchmark de Performance - DP vs Force Brute',
                 fontsize=14, fontweight='bold', color=C['text'])

    def style_ax(ax):
        ax.set_facecolor(C['card']); ax.grid(True, alpha=0.2)
        ax.tick_params(colors=C['muted'])
        for sp in ax.spines.values(): sp.set_edgecolor(C['border'])

    axes[0].plot(ns_dp, t_dp, color=C['accent'], linewidth=2.5, marker='o', markersize=4)
    axes[0].fill_between(ns_dp, t_dp, alpha=0.15, color=C['accent'])
    axes[0].set_xlabel("Nb objets (n)", color=C['muted'])
    axes[0].set_ylabel('Temps (ms)', color=C['muted'])
    axes[0].set_title(f'DP seule (W={W})', color=C['text']); style_ax(axes[0])

    axes[1].plot(ns_cmp, t_dp_cmp, color=C['green'],   linewidth=2.5, marker='o', markersize=4, label='DP O(nxW)')
    axes[1].plot(ns_cmp, t_fb_cmp, color=C['accent2'], linewidth=2.5, marker='s', markersize=4, label='Force Brute O(2^n)')
    axes[1].set_xlabel("Nb objets (n)", color=C['muted'])
    axes[1].set_ylabel('Temps (ms)', color=C['muted'])
    axes[1].set_title('DP vs Force Brute', color=C['text'])
    axes[1].legend(fontsize=10, framealpha=0.2, labelcolor=C['text']); style_ax(axes[1])

    axes[2].plot(Ws, t_W, color=C['yellow'], linewidth=2.5, marker='o', markersize=4)
    axes[2].fill_between(Ws, t_W, alpha=0.15, color=C['yellow'])
    axes[2].set_xlabel('Capacite W', color=C['muted'])
    axes[2].set_ylabel('Temps (ms)', color=C['muted'])
    axes[2].set_title('DP en fonction de W (n=20 fixe)', color=C['text']); style_ax(axes[2])

    plt.tight_layout(); plt.show(); print('Benchmark termine !')

# ── Tests ────────────────────────────────────────────────────
def render_tests():
    tests = [
        ('Exemple classique',    [2,3,4,5],         [3,4,5,6],         7,  9),
        ('Exemple predefini',    [2,3,4,5],         [3,4,5,6],         5,  7),
        ('Capacite zero',        [1,2,3],            [4,5,6],           0,  0),
        ('Tous trop lourds',     [10,20,30],         [5,6,7],           5,  0),
        ('Un objet - pris',      [3],                [9],               3,  9),
        ('Un objet - trop lourd',[5],                [9],               3,  0),
        ('Tous identiques',      [2,2,2],            [3,3,3],           4,  6),
        ('Capacite exacte',      [3,4],              [7,8],             7, 15),
        ('Poids unitaires',      [1,1,1,1],          [1,2,3,4],         3,  9),
        ('Grand n',              list(range(1,21)),  list(range(10,30)),50, None),
    ]
    nb_ok = 0; rows = ''
    for desc, p, v, cap, exp in tests:
        try:
            val,_,_,_ = sac_a_dos_01(p, v, cap)
            val_1d    = sac_a_dos_1D(p, v, cap)
            if exp is None:
                ok = (val == val_1d)
                result_txt = 'OK (execution)' if ok else f'2D={val} != 1D={val_1d}'
            else:
                ok = (val == exp and val_1d == exp)
                result_txt = f'2D={val} | 1D={val_1d}' if ok else f'Attendu {exp} | 2D={val} | 1D={val_1d}'
        except Exception as e:
            ok = False; result_txt = f'Erreur: {e}'
        nb_ok += int(ok)
        s = (f'<span class="t-ok">{result_txt}</span>' if ok
             else f'<span class="t-fail">{result_txt}</span>')
        exp_txt = str(exp) if exp is not None else '?'
        rows += (f'<div class="sda-trow">'
                 f'<span class="sda-tname">{desc}</span>'
                 f'<span class="sda-tparam">n={len(p)}, W={cap}, att={exp_txt}</span>'
                 f'{s}</div>')
    color   = '#43d98c' if nb_ok == len(tests) else '#ffd166'
    summary = (f'<div style="margin-top:12px;font-family:monospace;font-size:13px;color:{color}">'
               f'{nb_ok}/{len(tests)} tests reussis</div>')
    display(HTML(f'<div class="sda-app"><div class="sda-section">'
                 f'<div class="sda-section-title">Tests de Validation ({nb_ok}/{len(tests)})</div>'
                 f'{rows}{summary}</div></div>'))

# ── Interface principale ─────────────────────────────────────
def lancer_interface():
    inject_css()
    display(HTML("""
    <div class="sda-app">
    <div class="sda-hero">
      <div class="sda-hero-title">Sac a Dos 0/1 - Programmation Dynamique</div>
      <div class="sda-hero-sub">Master ISI - Aya SABRI - Basma HAIMER - Saida ELMASSAOUDI - Encadrant : MR. KASMI</div>
      <span class="sda-badge">Programmation Dynamique</span>
      <span class="sda-badge">O(n x W)</span>
      <span class="sda-badge">Python 3</span>
      <span class="sda-badge">Colab Interactive</span>
    </div></div>
    """))

    sl = {'description_width': '130px'}
    lf = widgets.Layout(width='98%')
    lh = widgets.Layout(width='48%')

    mode_btn  = widgets.ToggleButtons(
        options=['Manuel','Predefini','Aleatoire'], description='Mode :',
        style={'description_width':'60px','button_width':'130px'},
        layout=widgets.Layout(margin='10px 0'))
    noms_w    = widgets.Text(value='Livre,Radio,Telephone,Ordinateur', description='Noms :',    style=sl, layout=lf)
    poids_w   = widgets.Text(value='2,3,4,5',  description='Poids :',   style=sl, layout=lf)
    valeurs_w = widgets.Text(value='3,4,5,6',  description='Valeurs :', style=sl, layout=lf)
    cap_w     = widgets.BoundedIntText(value=7,  min=1, max=500, description='Capacite :',  style=sl, layout=lh)
    n_rand    = widgets.BoundedIntText(value=5,  min=2, max=20,  description='Nb objets :', style=sl, layout=lh)
    cap_rand  = widgets.BoundedIntText(value=15, min=5, max=200, description='Capacite :',  style=sl, layout=lh)

    box_manual = widgets.VBox([noms_w, poids_w, valeurs_w, cap_w])
    box_predef = widgets.HTML('<div style="color:#8b8fa8;font-size:13px;padding:8px 0">'
                              'Exemple: Livre(2,3), Radio(3,4), Telephone(4,5), Ordinateur(5,6), capacite=7</div>')
    box_rand   = widgets.VBox([n_rand, cap_rand])
    input_area = widgets.VBox([box_manual])

    def on_mode(change):
        if   change['new'] == 'Manuel':    input_area.children = [box_manual]
        elif change['new'] == 'Predefini': input_area.children = [box_predef]
        else:                              input_area.children = [box_rand]
    mode_btn.observe(on_mode, names='value')

    opt_charts = widgets.Checkbox(value=True,  description='Graphiques',   indent=False)
    opt_step   = widgets.Checkbox(value=True,  description='Pas-a-pas',    indent=False)
    opt_back   = widgets.Checkbox(value=False, description='Backtracking', indent=False)
    opt_bench  = widgets.Checkbox(value=False, description='Benchmark',    indent=False)
    opt_tests  = widgets.Checkbox(value=False, description='Tests',        indent=False)
    opts = widgets.HBox([opt_charts, opt_step, opt_back, opt_bench, opt_tests],
                        layout=widgets.Layout(flex_wrap='wrap', margin='8px 0'))

    btn_run  = widgets.Button(description='Lancer la resolution', button_style='primary',
                              layout=widgets.Layout(width='220px', height='40px'),
                              style={'font_weight': 'bold'})
    out_main = widgets.Output()

    def on_run(b):
        with out_main:
            clear_output(wait=True)
            try:
                mode = mode_btn.value
                if mode == 'Manuel':
                    noms     = [s.strip() for s in noms_w.value.split(',')]
                    poids    = [int(s.strip()) for s in poids_w.value.split(',')]
                    valeurs  = [int(s.strip()) for s in valeurs_w.value.split(',')]
                    capacite = cap_w.value
                elif mode == 'Predefini':
                    noms=['Livre','Radio','Telephone','Ordinateur']
                    poids=[2,3,4,5]; valeurs=[3,4,5,6]; capacite=7
                else:
                    n = n_rand.value; capacite = cap_rand.value
                    poids   = [random.randint(1, max(1, capacite//3)) for _ in range(n)]
                    valeurs = [random.randint(1, 20) for _ in range(n)]
                    noms    = [f'Obj{i+1}' for i in range(n)]
                    print(f'Generes : poids={poids}, valeurs={valeurs}')

                inject_css()
                val, sel_idx, pt, DP = render_solution(poids, valeurs, capacite, noms)
                if opt_charts.value:
                    render_charts(poids, valeurs, sel_idx, capacite, noms)
                if opt_step.value:
                    display(HTML('<div class="sda-section" style="margin-top:16px">'
                                 '<div class="sda-section-title">Visualisation Pas-a-Pas</div></div>'))
                    render_stepbystep(poids, valeurs, capacite, noms)
                if opt_back.value:
                    display(HTML('<div class="sda-section" style="margin-top:16px">'
                                 '<div class="sda-section-title">Heatmap Backtracking</div></div>'))
                    render_backtracking(poids, valeurs, capacite, noms)
                if opt_bench.value:
                    render_benchmark(W=capacite)
                if opt_tests.value:
                    render_tests()
            except Exception as e:
                display(HTML(f'<div style="color:#ff6584;font-family:monospace;padding:12px;'
                             f'background:#1a0a0f;border-radius:8px;border:1px solid #ff6584">'
                             f'Erreur : {e}</div>'))

    btn_run.on_click(on_run)
    display(widgets.VBox([
        widgets.HTML('<div style="color:#8b8fa8;font-size:13px;margin-bottom:6px">Mode de saisie</div>'),
        mode_btn, input_area,
        widgets.HTML('<div style="color:#8b8fa8;font-size:13px;margin:10px 0 4px">Options</div>'),
        opts, btn_run, out_main
    ], layout=widgets.Layout(padding='4px 0')))

print('Moteur et interface charges !')
print('Executez la cellule ci-dessous pour lancer l application.')
```
